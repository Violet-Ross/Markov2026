```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
a = 0.9
lam_f = 1000
lam_s = 10
```

```python
times = []
for i in range(10**5):
    X = np.random.binomial(1,a)
    U = np.random.uniform(low = 0, high = 1, size = 1).item()
    if X == 1:
        T = -math.log(U)/lam_f
    else:
        T = -math.log(U)/lam_s
    times.append(T)


def f(t):
    y = a * lam_f * (math.e ** (-lam_f * t)) + (1 - a) * lam_s * (math.e ** (-lam_s * t))
    return(y)

t = np.linspace(0, 2, 100)

y = f(t)

plt.hist(times, bins=30, density=True, edgecolor = 'black')
plt.plot(t, y)
plt.title("Problem 4c")
plt.xlabel("$t$")
plt.ylabel("$f(t)$")
plt.yscale("log")
plt.savefig("hw2_p4.png")
```

```python
print(f"Empirical mean is {sum(times) / len(times)}")
print(f"Prob > 50 is {sum(np.array(times) > 0.05) / len(times)}")
```
