```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

# Part b

```python
z_vals = []
for i in range(10**5):
    U = np.random.uniform(low = 0, high = 1, size = 1).item()
    Z = 1 - (1 - U) ** (1/2)
    z_vals.append(Z)


plt.hist(z_vals, bins=30, density=True, edgecolor = 'black')
plt.title("Problem 5b")
plt.xlabel("$z$")
plt.ylabel("$density$")
```

```python
sum(z_vals) / len(z_vals)
```

# Part c

```python
R = 10**3
N = 10**4

C = np.ones(R, dtype=int)

for n in range(3, N):
    p = C / n
    C += np.random.binomial(1, p)

z = C / N

x = np.linspace(0, 1, 1000)
h = 2 * (1 - x)

plt.hist(z, bins=30, density=True, edgecolor='black')
plt.plot(x, h)

plt.xlabel("$z$")
plt.ylabel("Density")
plt.title("Problem 5c")
plt.savefig("hw1_p5c.png")

print(f"Empirical mean: {np.mean(z)}")
print(f"Empirical SD/mean: {np.std(z) / np.mean(z)}")

print(f"Smallest core: {np.min(C)}")
print(f"Largest core: {np.max(C)}")
```
