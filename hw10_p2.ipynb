```python
import math
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
```

```python
t = 0
n = 0
lambdaa = 12
mu = 3
rho = lambdaa / mu
T = 500

time_in_state = {} 

while t < T:
    expo_rate = lambdaa + n * mu
    waiting_time = np.random.exponential(1 / expo_rate)  

    waiting_time = min(waiting_time, T - t)
    time_in_state[n] = time_in_state.get(n, 0) + waiting_time
    t += waiting_time
    if t >= T:
        break

    up_prob = lambdaa / (lambdaa + n * mu)
    move_up = np.random.binomial(1, up_prob)

    if move_up == 1:
        n += 1
    else:
        n -= 1
```

```python
K = max(time_in_state) + 1
frac = np.array([time_in_state.get(k, 0) / T for k in range(K)])
pois = np.array([math.exp(-rho) * rho**k / math.factorial(k) for k in range(K)])

x = np.arange(K)
w = 0.4
fig, ax = plt.subplots(figsize=(8, 4.5))   # new figure every run
ax.bar(x - w / 2, frac, w, label="Simulated fraction of time")
ax.bar(x + w / 2, pois, w, label="Poisson(4) pmf")
ax.set_xlabel("n")
ax.set_ylabel("probability")
ax.legend()
fig.savefig("p2.png", dpi=300)
plt.close(fig)

diffs = np.abs(frac - pois)
print("Largest difference: %.4f at n = %d" % (diffs.max(), diffs.argmax()))

time_avg = sum(k * s for k, s in time_in_state.items()) / T
print("Time average of X_t: %.4f" % (time_avg))

cdf = 0
c = 0
while True:
    cdf += math.exp(-rho) * rho**c / math.factorial(c)
    if 1 - cdf <= 0.01:
        break
    c += 1
print("Smallest fleet size c: %d " % (c))
```
