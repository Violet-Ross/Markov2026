```python
N = [10**1, 10**2, 10**3, 10**4, 10**5, 10**7]
P = []
for n in N:
    X = np.random.uniform(0,1, n)
    Y = np.random.uniform(0,1, n)
    Z = np.random.uniform(0,1, n)

    p = (sum((X**2 + Y**2 < Z) & (Z**2 > X*Y)) / len(X)).item()

    P.append(p)

plt.scatter(x = N, y = P)
plt.axhline(y = 0.375)
plt.xscale('log')
plt.xlabel("N")
plt.ylabel("probability")
```
