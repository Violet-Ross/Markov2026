```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
## accept-reject
N = 10**4
lam = 1/2
accepted = []
proposal_count = 0

start = time.time()
while len(accepted) < N:
    proposal_count += 1
    X = np.random.exponential(1/lam, 1).item()
    U = np.random.uniform(low = 0, high = 1, size = 1).item()
    if(U < (np.e / 2) * X * (np.e ** (-X / 2))):
        accepted.append(X)
end = time.time()
mean_time = (end - start) / N

## actual f(x)
def f(x):
    y = x * np.e ** (-x)
    return(y)

x = np.linspace(0.01, 12, N)

y = f(x)

plt.hist(accepted, bins=30, density=True, edgecolor = 'black')
plt.plot(x, y)
plt.title("Problem 3c with $\\lambda = 0.5$")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.savefig("hw2_p3a.png")

acceptance_frac = len(accepted) / proposal_count
print(f"Empirical acceptance fraction is {acceptance_frac} compared to a theoretical value of {1 / (4 / np.e)} ")

print(f"Mean time per accepted sample: {mean_time:.2e} seconds")
```
